```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv('IPL_Matches_Data_2008_2026.csv')
sns.set_style('whitegrid')

print('कुल मैच और कॉलम:', df.shape)
display(df.head())

plt.figure(figsize=(10, 5))
top_winners = df['winner'].value_counts().head(10)
sns.barplot(
    x=top_winners.values,
    y=top_winners.index,
    hue=top_winners.index,
    palette='viridis',
    legend=False,
)
plt.title('Top 10 Most Successful IPL Teams (2008-2026)', fontsize=14)
plt.xlabel('Total Wins')
plt.ylabel('Teams')
plt.show()

plt.figure(figsize=(6, 6))
toss_counts = df['toss_decision'].value_counts()
plt.pie(
    toss_counts,
    labels=toss_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=['#4CAF50', '#2196F3'],
)
plt.title('Toss Decision Distribution (Bat vs Field)', fontsize=14)
plt.show()

plt.figure(figsize=(12, 5))
sns.countplot(
    x='season', data=df, hue='season', palette='coolwarm', legend=False
)
plt.title('Matches Played per Season (2008-2026)', fontsize=14)
plt.xticks(rotation=45)
plt.xlabel('Season')
plt.ylabel('Number of Matches')
plt.show()

plt.figure(figsize=(10, 5))
top_players = df['player_of_match'].value_counts().head(10)
sns.barplot(
    x=top_players.values,
    y=top_players.index,
    hue=top_players.index,
    palette='magma',
    legend=False,
)
plt.title('Top 10 Player of the Match Winners', fontsize=14)
plt.xlabel('Awards Count')
plt.ylabel('Player Name')
plt.show()
```
